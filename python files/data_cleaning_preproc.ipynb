```python
# Create a function to calculate the final price of a product after applying a discount.
def final_price(price,discount):
    final=price-(price*discount/100)
    return final
result=final_price(20000,15)
print("Final Price on 15% Discount:-",result)
```

```python
#Create a function to calculate the total bill of a customer based on quantity and price.
def Total_Bill(price,quantity):
    total=price*quantity
    return total
t=Total_Bill(500,3)
print("Total bill:-",t)
```

```python
#Create a function to classify customers into the following categories based on their age:
def classify(age):
    if age<=0:
        return "Invalid age"
    elif age>0 and age<12:
        return "Child"
    elif age<=19:
        return "Teenager"
    elif age>=20 and age<=59:
        return "Adult"
    elif age>=60:
        return "Senior Citizen"
    else:
        return("Please enter proper age!")
print("This age falls under category of :",classify(56))
```

```python
#Create a function that accepts a list Of product prices and returns the highest-priced product.
def highest_priced_prod(df):
    index=df["Price"].idxmax()
    return df.loc[index,['Product_Name','Price']]
print("Highest Priced Product and its Price:")
print("-"*25)
print(highest_priced_prod(df))
```

```python
#Create a function that accepts customer spending and customers as:
# Low Spender
# Medium Spender
# High Spender
def cust_seg(spending):
    if spending<=1000:
        return "Low Spender"
    elif spending>=5000:
        return "Medium Spender"
    else:
        return "High Spender"

cust_seg(5500)
```

```python
def calculate_average_rating(ratings):
    if len(ratings) == 0:
        return 0

    average = sum(ratings) / len(ratings)
    return average

customer_ratings = [4, 5, 3, 5, 4]

result = calculate_average_rating(customer_ratings)

print("Average Customer Rating:", result)
```

```python
def festive_discount_eligibility(purchase_amount, is_loyal_customer):
    if purchase_amount >= 5000 and is_loyal_customer:
        return "Eligible for Additional Festive Discount"
    else:
        return "Not Eligible for Additional Festive Discount"
print(festive_discount_eligibility(6000, True))
print(festive_discount_eligibility(3000, False))
```

```python
#1 Create an array of 15 product prices
prices=np.array(df["Price"].head(15))
print("Product Prices:",prices)
```

```python
#2 Find the maximum and minimum product prices.
Max=np.max(df["Price"])
print("Maximum Product Price is ",Max)
min=np.min(df["Price"])
print("Minimum Product Price is ",min)
```

```python
#3 Calculate mean, median and standard deviation of product prices.
mean=np.mean(df["Price"])
median=np.median(df["Price"])
std_=np.std(df["Price"])
print("Mean of Price-",mean)
print("Median of Price-",median)
print("Standard Deviation of Price -",std_)
```

```python
# 4 Identify all prices greater than the average product price.
avg_price=np.mean(df["Price"])
print("Average Product Price:",avg_price)
print("\nPrices Greater than Average:")
print("-"*50)
print(df.loc[df["Price"] > avg_price,["Product_Name","Price"]])
```

```python
#7 Create an array containing daily sales for 1O days.
daily_sales=df.groupby('Order_Date')['Total_Bill'].sum()
sales_10_days=np.array(daily_sales.head(10))
print("Daily Sales for 10 Days-")
print(sales_10_days)
```

```python
#8 Calculate the total, average and maximum daily sales.
total_sales=np.sum(sales_10_days)
avg_sales=np.mean(sales_10_days)
mx_sales=np.max(sales_10_days)
print("Total daily sales:",total_sales)
print("Average sales:",avg_sales)
print("Maximum sales:",mx_sales)
```

```python
# 9 Create two arrays representing sales Of two different product categories and perform element-wise addition.
# first 10 dates
dates=daily_sales.head(10).index

#Calc Electronics sales
elec_sales=df[df["Category"]=="Electronics"].groupby('Order_Date')['Total_Bill'].sum()
elec_arr=np.array(elec_sales.reindex(dates,fill_value=0))

cloth_sales=df[df["Category"]=="Clothing"].groupby('Order_Date')['Total_Bill'].sum()
cloth_arr=np.array(cloth_sales.reindex(dates,fill_value=0))

print("Electronics sales:")
print(elec_arr)
print("Clothing sales:")
print(cloth_arr)
comb=elec_arr+cloth_arr
print("Element wise sales:")
print(comb)
```
