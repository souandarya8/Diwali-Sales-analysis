```python
#------------------------------------------Data Cleaning---------------------------------------------
#1. Check for missing values in every column.
df.isnull().sum()
```

```python
#2. Identify duplicate records.
df.duplicated().sum()
```

```python
#3. Convert the Order Date column into the appropriate date format.
df["Order_Date"]=pd.to_datetime(df["Order_Date"])
print(df["Order_Date"])
```

```python
# 4. Create a new column named Total Amount representing the amount payable after discount.
df["Total_Amount"]=df["Total_Bill"]-(df["Total_Bill"]*df["Discount"]/100)
disc = df[["Total_Bill","Discount","Total_Amount"]]
disc.head(10)
```

```python
df.columns
```

```python
#---------------------------------------------Data Analysis----------------------------------------
# 1. Find the top 5 most purchased products.
most_purchased_products = df.groupby('Product_Name')['Quantity'].sum().sort_values(ascending=False).head(5)
most_purchased_products
```

```python
# 2. Calculate the total revenue generated by the company.
Total_revenue=df["Total_Amount"].sum()
print("Total Revenue:",Total_revenue)
```

```python
# 3. Identify the city generating the highest revenue.
highest_revenue=df.groupby("City")["Total_Amount"].sum().sort_values(ascending=False).head(1)
highest_revenue
```

```python
# 4. Find the most preferred payment method.
preferred_pay=df.groupby("Payment_Mode")["Payment_Mode"].count()
print("The Most prefferred payment method is -",preferred_pay.idxmax())
```

```python
#5. Calculate the average customer rating.
avr_rating=df["Rating"].mean()
print("Average Customer rating-",avr_rating)
```

```python
# 6. Find the customer who has spent the maximum amount.
max_spent=df.groupby("Customer_ID")["Total_Amount"].sum()
cust=max_spent.idxmax()
amt=max_spent.max()
cust_name=df.loc[df["Customer_ID"]==cust,"Customer_Name"].iloc[0]
print("Max spent customer is ",cust_name)
print("Total amount spent - rs.",amt)
```

```python
# 7. Calculate city-wise total sales.
city_wise=df.groupby("City")["Total_revenue"].sum().sort_values(ascending=False)
print("City wise sales")
print(city_wise)
```

```python
#8.Calculate category-wise revenue
category_wise=df.groupby("Category")["Total_revenue"].sum().sort_values(ascending=False)
print("category wise sales")
print(category_wise)
```

```python
# 9. Find the average spending Of customers.
cust_spending=df.groupby("Customer_ID")["Total_Amount"].sum()
avg_spending=cust_spending.mean()
print("Average customer spending is Rs.",round(avg_spending),2)
```

```python
#10. Identify customers who have spent more than 5,OOO
c=df.groupby("Customer_Name")["Total_Amount"].sum()
high_spender=round(c[c>5000]),2
print("Customer who have spend more than 5k",high_spender)
```

```python
#11. Find the product category having the highest quantity sold.
category_qty=df.groupby("Category")["Quantity"].sum()
high_qty=category_qty.idxmax()
print(f"{high_qty} having the highest quantity sold")
```

```python
# 12. Calculate the total discount offered by the company.
df["Discount_Amt"]=(df["Quantity"]*df["Price"]*df["Discount"]/100)
total_disc=df["Discount_Amt"].sum()
print("Total Discount offered by the company is-",total_disc)
```

```python
# 13. Find the number Of orders placed on each date
order_per_dt=df.groupby("Order_Date").size().reset_index(name="No. of Orders")
order_per_dt
```

```python
# 14. Identify the top 5 customers based on their total spending.
total_spending=df.groupby("Customer_Name")["Total_Bill"].sum().sort_values(ascending=False).head(5)
total_spending.reset_index(name="Top 5 customers")
```

```python
# 15. Find the average age Of customers purchasing from each product category.
prod_cat=df.groupby("Category")["Age"].mean()
round(prod_cat.reset_index(name="Customer age"))
```

```python
#16. Identify the product having the lowest sales revenue.
low_sales=df.groupby("Product_Name")["Total_Bill"].sum()
print("Product having lowest sales-",low_sales.idxmin())
```

```python
# Create discount groups
df["Discount_Status"] = df["Discount"].apply(
    lambda x: "Discount Applied" if x > 0 else "No Discount"
)

# Compare average quantity purchased
discount_analysis = df.groupby("Discount_Status")["Quantity"].mean()

print(discount_analysis)
```
