```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
file="Dataset_.xlsx"
df=pd.read_excel(file)
```

```python
# Calculate total revenue for each category
category_revenue = df.groupby("Category")["Total_Bill"].sum().reset_index()

# Create bar chart
sns.barplot(data=category_revenue, x="Category", y="Total_Bill",color="cyan")
plt.title("Category-wise Revenue")
plt.xlabel("Product Category")
plt.ylabel("Total Revenue")
plt.show()
```

```python
# 2. Create a pie chart showing payment method distribution.
pay_counts=df["Payment_Mode"].value_counts()
plt.pie(pay_counts,labels=pay_counts.index,startangle=90,autopct="%.0f%%")
plt.title("Payment method distribution")
plt.show()
```

```python
# 3. Create a histogram representing customer age distribution.
cust_age=df["Age"].value_counts()
plt.figure()
sns.histplot(data=df,x="Age",bins=10,kde=True)
plt.title("Customer Age Distribution")
plt.xlabel("Customer Age")
plt.ylabel("Number of Customers")
plt.show()
```

```python
#4. Create a count plot showing the number of orders placed by customers from different cities
plt.Figure()
sns.countplot(data=df,x="City")
plt.title("The number of orders by customer city wise")
plt.xlabel("Cities")
plt.ylabel("Number of Orders")
plt.show()
```

```python
# 5. Create a box plot representing the distribution of product prices.
plt.figure(figsize=(4,4))
sns.boxplot(data=df,y="Price")
plt.title("distribution of product prices")
plt.ylabel("Product Price")
plt.show()
```

```python
# 6. Cleate a line chan showing daily sales trends.
daily_sales=df.groupby("Order_Date")["Total_Bill"].sum().reset_index()
plt.figure(figsize=(16,4))
sns.lineplot(data=daily_sales,x="Order_Date",y="Total_Bill",markers="o")
plt.title("Daily sales Trends")
plt.xlabel("Order Date")
plt.ylabel("Sales")
plt.show()
```

```python
# 7. Create a heatmap showing the correlation between numerical columns.
numeric_df=df.select_dtypes(include="number")
corr=numeric_df.corr()
plt.figure(figsize=(5,5))
sns.heatmap(corr,annot=True,cmap="coolwarm",fmt=".2f")
plt.title("Correlation between numerical data")
plt.show()
```

```python
# 8. Create a scatter plot showing the relationship between quantity purchased and total amt
plt.figure(figsize=(4,4))
sns.scatterplot(data=df,x="Quantity",y="Total_Bill")
plt.title("Relation between Quantity and Total amount")
plt.xlabel("Quantity")
plt.ylabel("Total Bill")
plt.show()
```

```python
# 9. Create a bar chart displaying the top 5 most purchased products.
most_purchased_products = df.groupby('Product_Name')['Quantity'].sum().sort_values(ascending=False).head(5).reset_index()
most_purchased_products
plt.figure(figsize=(8,6))
sns.barplot(data=most_purchased_products,x="Product_Name",y="Quantity")
plt.title("The Top 5 most purchased Products")
plt.xlabel("Product name")
plt.ylabel("Quantity sold")
plt.show()
```

```python
#10. Create any additional visualization Of your choice to represent an interesting insight from your dataset.
plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x="Category",y="Price",color="Brown")
plt.title("Product Price Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Product Price")
plt.show()
```
